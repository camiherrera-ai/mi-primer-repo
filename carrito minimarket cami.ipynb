### Punto 2. Construcción de funciones

En esta sección se deben construir todas las funciones utilizando `return`.

Las funciones deben encargarse de procesar la información y entregar un resultado. No deben modificar directamente los elementos de la interfaz, como `widget_salida`.

Esto permite separar la lógica del programa de la interfaz:

- Las funciones realizan cálculos, búsquedas o modificaciones en los datos.
- Las acciones de los botones reciben el resultado y lo muestran en la interfaz.

Por ejemplo:

```python
def calcular_total(carrito):
    total = 0

    for producto in carrito:
        total = total + producto['precio'] * producto['cantidad']

    return total
```

Después, la función que conecta el botón puede mostrar ese resultado:

```python
def accion_mostrar_total(boton):
    total = calcular_total(carrito)
    widget_salida.value = 'Total de compra: $' + str(total)
```

Trabajar con `return` hace que las funciones sean más ordenadas, reutilizables y fáciles de revisar. Además, una misma función puede ser utilizada por diferentes botones o partes del programa.

### Punto 4. Conexión de botones con las funciones

En esta sección se deben crear las acciones que conectan la interfaz con las funciones construidas en el punto 2.

Estas acciones funcionan como un punto de conexión entre los elementos visuales y la lógica del programa. Su tarea es:

1. Recibir la señal de que el usuario presionó un botón.
2. Obtener los valores ingresados en los widgets.
3. Llamar a la función correspondiente.
4. Recibir el resultado de `return`de cada función del punto 2.
5. Mostrar el resultado en la interfaz utilizando los widgets de salida.

La estructura general de una acción es:

```python
def nombre_de_la_accion(boton):
    resultado = funcion_correspondiente(datos)
    widget_salida.value = resultado
```

Finalmente, cada acción debe conectarse con su botón mediante el método `.on_click()`:

```python
widget_boton.on_click(nombre_de_la_accion)
```

De esta manera, las funciones se encargan de procesar los datos y las acciones se encargan de comunicar esos resultados a la interfaz.

````python
# Celda 4. Conexión de botones con las funciones

def accion_limpiar(boton):  # deja la interfaz como nueva
    widget_busqueda.value=""   # borra todo texto (campo vacio)
    widget_categoria.value="Todas"  #busca en todas las categorias.
    widget_cantidad.value= 1  # Restablece el contador de unidades para que vuelva a marcar 1.
    widget_codigo_quitar.value=""  #borra cualquier codigo previo en sistema.
    widget_mensaje.value=""  #borra los avisos en pantalla.
    widget_salida.value=""  # limpia la pantalla donde se muestran tablas de productos



def accion_buscar(boton):  #asistente de pasillo en busca de productos
    resultados=buscar_productos(inventario, widget_busqueda.value, widget_categoria.value)
    if resultados.empty:
      widget_salida.value= "No se encontraron productos"
    else:
      widget_salida.value = resultados.to_html(index=False)



def accion_agregar(boton): # agrega productos a la bolsa
  resultados = buscar_productos(inventario, widget_busqueda.value, widget_categoria.value)
  if len(resultados) == 0:
        widget_mensaje.value = 'Busca un producto antes de agregarlo.'
        return
  codigo = resultados.iloc[0]['codigo']
  widget_mensaje.value = agregar_al_carrito(inventario, carrito, codigo, widget_cantidad.value)
  widget_salida.value = mostrar_carrito(carrito)




def accion_mostrar(boton): # muestra el ticket
    widget_salida.value= mostrar_carrito(carrito)


def accion_quitar(boton):  # devuelve el prodcuto al estante!
    widget_mensaje.value = quitar_del_carrito(carrito,widget_codigo_quitar.value)  #lee el codigo, busca el prodcuto del carrito, elimina, guarda mensaje ej:Producto eliminado del carrito."
    widget_salida.value = mostrar_carrito(carrito) # vuelve con el carrito sin el prodcuto

def accion_vaciar(boton):  #el que cancela la compra / cliente se arrepiente de la compra
    carrito.clear() #Borra absolutamente todos los elementos que estaban guardados en la lista del carrito.
    widget_mensaje.value= "El carrito fue vaciado"
    widget_salida.value= mostrar_carrito(carrito) # actualiza la interfaz


widget_boton_buscar.on_click(accion_buscar)
widget_boton_agregar.on_click(accion_agregar)
widget_boton_mostrar.on_click(accion_mostrar)
widget_boton_quitar.on_click(accion_quitar)
widget_boton_vaciar.on_click(accion_vaciar)
widget_boton_limpiar.on_click(accion_limpiar)
````

- Al presionar **“Ver carrito”** cuando el carrito está vacío, debe informar que no hay productos agregados.

- Al presionar **“Vaciar carrito”**, deben eliminarse todos los productos y el total debe volver a cero.

- Al presionar **“Quitar producto”** con un código que no está en el carrito, debe informar que el producto no se encuentra en el carrito o que no existe.

- Al presionar **“Quitar producto”** con un producto existente, debe actualizarse el carrito y recalcularse el total.

- Al presionar **“Limpiar”**, deben borrarse los campos de la interfaz, pero el carrito debe conservar sus productos.
